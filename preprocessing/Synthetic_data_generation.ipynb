```python
import openai
import pandas as pd

# GPT API Key
openai.api_key = "YOUR_API_KEY"

def generate_reviews(prompt, max_responses=10, temperature=0.7):
    """
    Generate a list of reviews based on the prompt and temperature setting.

    Args:
    - prompt (str): The text prompt for GPT.
    - max_responses (int): Number of responses to generate.
    - temperature (float): Sampling temperature (controls randomness).

    Returns:
    - list: List of generated reviews.
    """
    responses = []
    for _ in range(max_responses):
        response = openai.Completion.create(
            engine="text-davinci-003",
            prompt=prompt,
            max_tokens=150,
            n=1,
            stop=None,
            temperature=temperature
        )
        responses.append(response["choices"][0]["text"].strip())
    return responses

# Example prompts for AI-generated and human-written reviews
ai_prompt = "Write a fake AI-generated product review."
human_prompt = "Write a genuine human-authored product review."

# Temperatures to experiment with
temperatures = [0.2, 0.5, 0.7, 1.0]
max_responses_per_temp = 100

# Loop through each temperature and generate datasets
for temp in temperatures:
    print(f"\nGenerating dataset for temperature = {temp}...")

    # Generate AI-generated reviews
    ai_reviews = generate_reviews(ai_prompt, max_responses=max_responses_per_temp, temperature=temp)

    # Generate human-authored reviews
    human_reviews = generate_reviews(human_prompt, max_responses=max_responses_per_temp, temperature=temp)

    # Create a dataset
    data = pd.DataFrame({
        "text": ai_reviews + human_reviews,
        "label": [0] * len(ai_reviews) + [1] * len(human_reviews)
    })

    # Save the dataset with a filename indicating the temperature
    filename = f"generated_reviews_temp_{temp}.csv"
    data.to_csv(filename, index=False)
    print(f"Dataset saved to '{filename}'.")
```
